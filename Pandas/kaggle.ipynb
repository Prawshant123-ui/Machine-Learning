```python

import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv("train.csv")
df_original = df.copy()

# 2. Inspect the data
print(df.head())
print("Shape:", df.shape)
df.info()
print(df.describe(include="all"))

# 3. Check duplicates and missing values
print("Duplicate rows:", df.duplicated().sum())
print(df.isna().sum())

# 4. Handle missing values
df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)

df["Cabin"] = df["Cabin"].fillna("Unknown")

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median())

# 5. Clean data types and formatting
df["Fare"] = df["Fare"].round(2)
df["Sex"] = df["Sex"].str.capitalize()

# 6. Rename columns
df = df.rename(columns={
    "PassengerId": "Passenger_id",
    "Pclass": "P_class",
    "SibSp": "Sib_sp"
})

# 7. Clean Ticket without removing meaningful information
df["Ticket"] = df["Ticket"].str.strip()

# 8. Optional: create a readable survival label
df["Survived_label"] = df["Survived"].map({
    1: "Yes",
    0: "No"
})

# 9. Validate the cleaned data
print("Final shape:", df.shape)
print("Missing values:")
print(df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())

# 10. Export without saving the DataFrame index
df.to_csv("Final_good.csv", index=False)

df
```
