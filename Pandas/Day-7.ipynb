Sorting, Indexing & Reshaping

```python
import pandas as pd

df = pd.DataFrame({
    "name":  ["Ram", "Sita", "Hari", "Gita"],
    "city":  ["Pokhara", "Kathmandu", "Chitwan", "Kathmandu"],
    "marks": [78, 92, 85, 92]
})
```

Sort

```python
# df.sort_values()--> sort by column values
result=df.sort_values("marks")
result

# Highest first, ties broken by name A→Z
result1=df.sort_values(["marks", "name"], ascending=[False, True])
result1

#df.sort_index(): sort by index labels
df.sort_values("marks").sort_index()   # back to 0, 1, 2, 3 order
```

Index control

```python
#df.set_index(): turn a column into the index
d = df.set_index("name")
print(d)

#df.reset_index(): move the index back to a column
d.reset_index()              # "name" becomes a column again, index is 0..n
d.reset_index(drop=True)     # throw the old index away instead

#df.reindex(): force the index to a specific set of labels
d.reindex(["Ram", "Hari", "Binod"])
```
